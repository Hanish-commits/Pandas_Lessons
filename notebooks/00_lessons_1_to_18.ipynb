# Lesson 6 — Selecting Columns

The most common way to select a column is:

```python
df["ColumnName"]
```

This returns a **Series**.

To select multiple columns:

```python
df[["Column1", "Column2"]]
```

This returns a **DataFrame**.

````python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ravi", "Neha", "Amit"],
    "Age": [30, 25, 28],
    "Salary": [55000, 42000, 38000],
    "City": ["Delhi", "Mumbai", "Pune"]
})

df["Name"]
````

````python
df[["Name", "Salary"]]
````

### Why are there two brackets?

```python
df["Name"]
```

asks for **one column** → Series.

```python
df[["Name", "Salary"]]
```

asks for **a list of columns** → DataFrame.

You can also do:

```python
columns_needed = ["Name", "City"]
df[columns_needed]
```

Dot notation such as `df.Name` can work, but bracket notation is safer and more consistent.

**Beginner tip:** Prefer `df["Column"]`.

# Lesson 7 — Selecting Rows

Pandas gives four important tools:

| Method | Main idea |
|---|---|
| `loc` | select using labels |
| `iloc` | select using positions |
| `at` | one cell using labels |
| `iat` | one cell using positions |

The most important distinction is:

```text
loc  → label
iloc → position
```

````python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ravi", "Neha", "Amit", "Priya"],
    "Age": [30, 25, 28, 35],
    "Salary": [55000, 42000, 38000, 72000]
}, index=["A", "B", "C", "D"])

df
````

````python
# loc: label-based
print(df.loc["B"])

# iloc: position-based
print(df.iloc[1])

# at: one cell by label
print(df.at["B", "Salary"])

# iat: one cell by position
print(df.iat[1, 2])
````

## Very important: slicing difference

`loc` includes the ending label:

```python
df.loc["A":"C"]
```

includes A, B and C.

`iloc` uses Python-style positional slicing:

```python
df.iloc[0:3]
```

includes positions 0, 1, 2 but not 3.

### Mental model

```text
loc  → "Give me the row with this label."
iloc → "Give me the row at this position."
at   → "Give me this one labelled cell."
iat  → "Give me this one positional cell."
```

**Common mistake:** Thinking `loc` always means row number. It means **label**.

# Lesson 8 — Filtering Data

Filtering means:

> "Show me only the rows that satisfy my condition."

Example:

```python
df[df["Age"] > 30]
```

This works because the condition creates a Boolean Series of `True` and `False`.

````python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ravi", "Neha", "Amit", "Priya", "Karan"],
    "Age": [30, 25, 28, 35, 24],
    "Salary": [55000, 42000, 38000, 72000, 35000],
    "Department": ["IT", "HR", "Finance", "IT", "HR"]
})

df[df["Age"] > 28]
````

## Multiple conditions

Use:

- `&` for AND
- `|` for OR
- `~` for NOT

Always put each condition inside parentheses.

````python
# AND
df[(df["Age"] > 25) & (df["Salary"] > 40000)]

# OR
df[(df["Department"] == "IT") | (df["Department"] == "HR")]

# NOT
df[~(df["Department"] == "HR")]
````

### Why not `and` / `or`?

Python's `and` and `or` expect one overall truth value.

A Pandas Series contains many Boolean values, so Pandas uses:

```text
&  |  ~
```

### `isin()`

```python
df[df["Department"].isin(["IT", "HR"])]
```

### `between()`

```python
df[df["Age"].between(25, 30)]
```

### `query()`

```python
df.query("Age > 28 and Salary > 40000")
```

**Common mistake:**

```python
df[df["Age"] > 25 & df["Salary"] > 40000]
```

Avoid this. Use parentheses around each condition.

# Lesson 10 — Adding, Modifying & Deleting Columns

# Lesson 12 — Missing Values

Real datasets often contain missing information.

Pandas commonly represents missing numeric data as `NaN`.

Important methods:

- `isna()`
- `isnull()`
- `notna()`
- `dropna()`
- `fillna()`
- `ffill()`
- `bfill()`

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ravi", "Neha", "Amit", "Priya", "Karan"],
    "Age": [30, np.nan, 28, 35, np.nan],
    "Salary": [55000, 42000, np.nan, 72000, 35000],
    "City": ["Delhi", np.nan, "Pune", "Mumbai", np.nan]
})

df
````

````python
# Find missing values
df.isna()

# Same idea
df.isnull()

# Non-missing values
df.notna()

# Missing count per column
df.isna().sum()

# Total missing cells
df.isna().sum().sum()
````

## `dropna()`

Remove rows containing missing values:

```python
df.dropna()
```

Remove rows where a particular column is missing:

```python
df.dropna(subset=["Salary"])
```

Remove columns containing missing values:

```python
df.dropna(axis=1)
```

````python
# Fill all missing values with a fixed value
filled = df.fillna("Unknown")
filled
````

For numeric columns, you can use a statistic such as the mean:

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

## Forward fill

```python
df.ffill()
```

Uses the previous available value.

## Backward fill

```python
df.bfill()
```

Uses the next available value.

### Important beginner point

`ffill()` only fills **actual missing values**.

If you already changed:

```text
NaN → "Unknown"
```

then `"Unknown"` is no longer missing, so `ffill()` has nothing to fill there.

### Mental model

```text
isna     → find missing
dropna   → remove missing
fillna   → replace missing
ffill    → previous value
bfill    → next value
```

# Lesson 14 — Descriptive Statistics

# Lesson 16 — Groupby & Aggregation

This is one of the most important Pandas concepts.

## The big idea

Suppose you have employee data and want:

> "What is the average salary for each department?"

You don't want one overall average. You want one answer **per group**.

That's what `groupby()` does.

### Mental model

```text
GROUPBY
   ↓
Split data into groups
   ↓
Calculate something for each group
   ↓
Combine the results
```

This is often described as:

**Split → Apply → Combine**

````python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ravi", "Neha", "Amit", "Priya", "Karan", "Simran", "Vikas", "Anjali"],
    "Department": ["IT", "HR", "Finance", "IT", "HR", "Finance", "IT", "HR"],
    "City": ["Pune", "Mumbai", "Delhi", "Bangalore", "Mumbai", "Delhi", "Pune", "Mumbai"],
    "Salary": [55000, 42000, 38000, 72000, 35000, 48000, 65000, 45000],
    "Age": [30, 25, 28, 35, 24, 31, 29, 26]
})

df
````

````python
df.groupby("Department")["Salary"].mean()
````

Read this from left to right:

```python
df.groupby("Department")["Salary"].mean()
```

1. `df` → start with DataFrame
2. `groupby("Department")` → create department groups
3. `["Salary"]` → focus on salary
4. `.mean()` → calculate average salary for each group

````python
df.groupby("Department")["Salary"].sum()
df.groupby("Department")["Salary"].count()
df.groupby("Department")["Salary"].min()
df.groupby("Department")["Salary"].max()
````

## Multiple aggregations

```python
df.groupby("Department")["Salary"].agg(["min", "max", "mean"])
```

### Named aggregations

This produces clearer column names:

````python
df.groupby("Department")["Salary"].agg(
    Minimum_Salary="min",
    Maximum_Salary="max",
    Average_Salary="mean"
)
````

## Multiple columns

```python
df.groupby("Department").agg({
    "Salary": ["mean", "max"],
    "Age": "mean"
})
```

````python
df.groupby("Department").agg({
    "Salary": ["mean", "max"],
    "Age": "mean"
})
````

## Group by more than one column

```python
df.groupby(["Department", "City"])["Salary"].mean()
```

This creates groups such as:

```text
IT + Pune
HR + Mumbai
Finance + Delhi
```

### Most important pattern

```python
df.groupby("GROUP_COLUMN")["VALUE_COLUMN"].AGGREGATION()
```

Examples:

```python
df.groupby("Department")["Salary"].mean()
df.groupby("Department")["Salary"].sum()
df.groupby("Department")["Salary"].max()
```

**Beginner tip:** `groupby()` does not automatically mean "average". You choose the aggregation.

# Lesson 17 — Combining DataFrames

There are three important ideas:

```text
concat → stack
merge  → match using keys/columns
join   → match using indexes
```

We will also go deep on LEFT and RIGHT merge because this is a common point of confusion.

## 17.1 `concat()` — Stack DataFrames

### Row-wise

```python
pd.concat([df1, df2], ignore_index=True)
```

`axis=0` is the default and means stack rows.

### Column-wise

```python
pd.concat([df1, df2], axis=1)
```

`axis=1` means place columns side by side.

````python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Ravi", "Neha"],
    "Salary": [55000, 42000]
})

df2 = pd.DataFrame({
    "Name": ["Amit", "Priya"],
    "Salary": [38000, 72000]
})

pd.concat([df1, df2], ignore_index=True)
````

## 17.2 `merge()` — Match related tables

Suppose we have employees and department information.

````python
employees = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104],
    "Name": ["Ravi", "Neha", "Amit", "Priya"],
    "Department_ID": [1, 2, 3, 5]
})

departments = pd.DataFrame({
    "Department_ID": [1, 2, 3, 4],
    "Department": ["IT", "HR", "Finance", "Marketing"]
})

employees
````

The matching key is:

```text
Department_ID
```

### Inner merge

Only matching IDs are kept:

```python
pd.merge(employees, departments, on="Department_ID", how="inner")
```

### LEFT merge

The LEFT DataFrame is the first DataFrame.

```python
pd.merge(employees, departments, on="Department_ID", how="left")
```

This means:

> **Keep every row from `employees`.**

Priya has Department_ID 5, but there is no 5 in the departments table. Priya is still kept, and the department becomes `NaN`.

### RIGHT merge

```python
pd.merge(employees, departments, on="Department_ID", how="right")
```

This means:

> **Keep every row from `departments`.**

Marketing has Department_ID 4 but no employee has 4. Marketing is still kept, and the employee fields become `NaN`.

### The easiest rule

```text
how="left"  → keep the FIRST DataFrame
how="right" → keep the SECOND DataFrame
```

### All four

```text
left  → everything from left
right → everything from right
inner → only matches
outer → everything from both
```

## 17.3 `join()` — Match using indexes

`join()` is primarily index-based.

Let's create DataFrames whose indexes match:

````python
employees = pd.DataFrame({
    "Name": ["Ravi", "Neha", "Amit", "Priya"]
}, index=[1, 2, 3, 4])

departments = pd.DataFrame({
    "Department": ["IT", "HR", "Finance", "Marketing"],
    "Location": ["Delhi", "Mumbai", "Pune", "Bangalore"]
}, index=[1, 2, 3, 4])

employees.join(departments)
````

Pandas matches:

```text
employees index    departments index
       1      ↔           1
       2      ↔           2
       3      ↔           3
       4      ↔           4
```

### Join types

```python
employees.join(departments)                 # left by default
employees.join(departments, how="left")
employees.join(departments, how="right")
employees.join(departments, how="inner")
employees.join(departments, how="outer")
```

### `merge()` vs `join()`

```text
merge → usually match columns/keys
join  → usually match indexes
```

If the matching key is a normal column such as `Department_ID`, `merge()` is usually the more natural choice.

### Final mental model

```text
CONCAT
  ↓
Stack DataFrames

MERGE
  ↓
Match related rows using columns/keys

JOIN
  ↓
Match related rows using indexes
```

# Lesson 18 — Reshaping Data

Reshaping means changing the **structure** of a DataFrame without changing the underlying information.

We will learn:

1. `pivot()`
2. `pivot_table()`
3. `melt()`
4. `transpose()` / `.T`

## Long vs Wide

Long:

```text
Month | City   | Sales
Jan   | Delhi  | 55000
Feb   | Delhi  | 60000
Jan   | Mumbai | 50000
```

Wide:

```text
Month | Delhi | Mumbai
Jan   | 55000 | 50000
Feb   | 60000 | ...
```

````python
import pandas as pd

df = pd.DataFrame({
    "Month": ["Feb", "Jan", "Mar", "Jan", "Mar", "Feb"],
    "City": ["Delhi", "Delhi", "Delhi", "Mumbai", "Mumbai", "Mumbai"],
    "Sales": [60000, 55000, 65000, 50000, 58000, 52000]
})

df
````

## 18.1 `pivot()` — Long → Wide

```python
df.pivot(
    index="Month",
    columns="City",
    values="Sales"
)
```

- `index` → what becomes rows
- `columns` → what becomes columns
- `values` → what fills the cells

````python
df.pivot(
    index="Month",
    columns="City",
    values="Sales"
)
````

### How to think about it

```text
pivot(
    rows,
    columns,
    values
)
```

It reorganizes the data.

### Important limitation

`pivot()` expects each combination of:

```text
index + columns
```

to be unique.

If the same Month + City occurs multiple times, Pandas does not know which value should occupy the cell.

## 18.2 `pivot_table()` — Long → Wide + Aggregation

Here is a dataset with duplicate Month + City combinations:

````python
sales = pd.DataFrame({
    "Month": ["Jan", "Jan", "Jan", "Feb", "Feb", "Feb"],
    "City": ["Delhi", "Delhi", "Mumbai", "Delhi", "Mumbai", "Mumbai"],
    "Sales": [55000, 60000, 50000, 65000, 52000, 58000]
})

sales
````

````python
sales.pivot_table(
    index="Month",
    columns="City",
    values="Sales",
    aggfunc="sum"
)
````

For January + Delhi:

```text
55000 + 60000 = 115000
```

For February + Mumbai:

```text
52000 + 58000 = 110000
```

You can also use:

```python
aggfunc="mean"
aggfunc="max"
aggfunc="min"
```

### Key difference

```text
pivot       → reshape, no aggregation
pivot_table → reshape + aggregation
```

````python
sales.pivot_table(
    index="Month",
    columns="City",
    values="Sales",
    aggfunc="mean"
)
````

## 18.3 `melt()` — Wide → Long

Start with wide data:

````python
wide_df = pd.DataFrame({
    "Name": ["Ravi", "Neha", "Amit"],
    "Math": [85, 90, 78],
    "Science": [88, 84, 92],
    "English": [80, 86, 79]
})

wide_df
````

````python
long_df = wide_df.melt(
    id_vars="Name",
    var_name="Subject",
    value_name="Marks"
)

long_df
````

### Understand the arguments

`id_vars="Name"`

→ Keep Name as the identifier.

`var_name="Subject"`

→ Turn the old column names (`Math`, `Science`, `English`) into values in a new column called Subject.

`value_name="Marks"`

→ Put the scores into a new column called Marks.

### Mental model

```text
melt()
  ↓
Wide → Long
```

It is useful when you have many subject/measurement columns and want them represented as rows.

## 18.4 `transpose()` — Rows ↔ Columns

Create a small DataFrame:

````python
df = pd.DataFrame({
    "Name": ["Ravi", "Neha", "Amit"],
    "Age": [30, 25, 28],
    "Salary": [55000, 42000, 38000]
})

df
````

````python
df.T
````

You can also write:

```python
df.transpose()
```

Both do the same thing.

### Mental model

```text
transpose
    ↓
rows become columns
columns become rows
```

Transpose is different from pivot because it simply flips the DataFrame's orientation rather than reorganizing it based on selected data columns.

## Common mistakes

### Mistake 1 — Confusing pivot and pivot_table

```text
pivot       → no aggregation
pivot_table → aggregation possible
```

### Mistake 2 — Forgetting the direction of melt

```text
melt = Wide → Long
```

### Mistake 3 — Thinking transpose and pivot are the same

```text
transpose → mechanically flips rows and columns
pivot     → reorganizes data using selected columns
```

### Mistake 4 — Duplicate combinations with pivot

If there are multiple rows for the same:

```text
Month + City
```

use `pivot_table()` when you need to aggregate those values.
